## API (Application Programming Interface)

![](./images/api.png)

### API Types (by Scope and Audience)

- **Public (Open) APIs:** Available to any external developer with minimal restrictions. They are designed to encourage third-party integrations and often require simple authentication like an API Key. Example: Google Maps API embedded in a local store's website. 
- **Partner APIs:** Shared exclusively with authorized, strategic business partners. They require stricter onboarding, specific license rights, and enhanced security workflows. Example: Amazon sharing inventory APIs with specific wholesale partners. 
- **Internal (Private) APIs:** Hidden completely from the public and kept strictly within an organization's network. They are used to bridge communication between internal development teams and different internal microservices. Example: A bank's internal system connecting its customer database to its mobile app dashboard.
Ex: local Fast API: http://127.0.0.1:8000
- **Code level/Library APIs:**  Invoked in the same process memory space. ex: json.loads(), genai.client()..

### Components of HTTP Web API

#### Endpoint (URI / URL):

    http:/api.example.com/sales/
    
    http:/api.example.com/service/

The network address pointing to a specific server resource or operation.

#### HTTP Method: 

- Get
- Post
- Put
- Patch
- Delete

#### Request Header

Key-value metadata pairs accompanying the request.

    Authorization: Bearer <API_KEY>

    Content-Type: application/json

    Accept: application/json

#### Request Body (Payload)

The raw data payload transmitted over the connection

    {"raw_text": "Alex Mercer...", "domain": "Data"}
