### 1. Git Global Identity Configuration
Ensure your commit authorship matches your registered GitHub account credentials:


Add these rules to your `.gitignore` file to prevent committing environments, caches, and `.env` credentials:

```text

venv/

__pycache__/

```

### 3. Initialize & First Commit
Stage workspace files and create the baseline commit:

```bash
# 1. Initialize Git repository
git init

# 2. Stage all project files
git add .

# 3. Create initial commit
git commit -m "Initial commit: workspace setup and environment config"
```

### 4. Connect Remote & Push Baseline
Establish the upstream link to GitHub and push `main`:

```bash
# 1. Ensure primary branch is named 'main'
git branch -M main

# 2. Link local repo to remote GitHub repo
git remote add origin https://github.com/<your-username>/<your-repo-name>.git

# 3. Push and track upstream
git push -u origin main
```

### Transitioning from Direct Pushes to Team Production Workflows

> **Core Engineering Rule:** `main` always represents clean, deployable code. Never write or push code directly to `main` in a team environment. All new work happens on an isolated branch.

### Step 1: Sync `main` and Create a Feature Branch
Ensure your local copy of `main` is current, then create an isolated workspace branch:

```bash
# 1. Switch to main and pull latest changes
git checkout main
git pull origin main

# 2. Create and switch to a dedicated feature branch
# Standard prefixes: feature/, fix/, chore/, refactor/
git checkout -b feature/add-git-branching-notes

# 3. Verify active branch (marked with * in green)
git branch
```

### Step 3: Publish Branch to GitHub
Branches created locally exist only on your machine until pushed upstream:

```bash

```

### Step 4: Open a Pull Request (PR) on GitHub
1. Open your repository on **GitHub**.
2. Locate the prompt banner: **`feature/add-prompt-runner had recent pushes... Compare & pull request`** and click it.
3. Verify branch direction: `base: main` $\leftarrow$ `compare: feature/add-prompt-runner`.
4. Complete the PR form:
   * **Title:** Concise, action-oriented summary (e.g., `feat: Add prompt runner validation utility`).
   * **Description:** Detail what changed and provide manual verification steps.
5. Click **Create Pull Request**.

### Step 5: Code Review & Merging
1. Navigate to the **Files changed** tab to review additions (green) and deletions (red).
2. Leave constructive inline review comments or submit an **Approve** review.
3. Select **Merge pull request** $\to$ **Confirm merge**.
4. Click **Delete branch** on GitHub to keep remote branch references tidy.

### Step 6: Local Cleanup & Resyncing `main`
Return to local `main`, pull the newly integrated commits, and remove the stale local feature branch:

```bash

git pull origin main

git branch -d feature/add-prompt-runner
```

# 🌐 Real-World Branching Strategies & Taxonomy
### How High-Performance Engineering Teams Manage Code

In enterprise development, branching models dictate how teams avoid merge collisions, enforce quality gates, and safely deliver to production.

---

### 1. Two Primary Branch Lifecycles
* **Permanent (Core) Branches:** Never deleted. Represent fixed stages of your software (e.g., `main`, `develop`).
* **Temporary (Supporting) Branches:** Short-lived branches created for a specific task and deleted immediately once merged back upstream.

## 🏛️ Permanent Branches

* **`main` (or `master`):**
  * **Role:** Production-ready, fully verified, deployable code.
  * **Rule:** Direct pushes are blocked; all code enters via reviewed, passing Pull Requests.
* **`develop`:**
  * **Role:** The integration branch used in models like **Git Flow**.
  * **Rule:** Features branch off `develop` and merge back into `develop`. Once stabilized, code is cut into a release branch before reaching `main`.

## 🗂️ Temporary Branches Cheat Sheet

| Branch Prefix | When to Use | Typical Base Branch | Merges Back Into |
| :--- | :--- | :--- | :--- |
| **`feature/`** (or `feat/`) | Adding new capabilities or endpoints | `main` / `develop` | `main` / `develop` |
| **`bugfix/`** (or `fix/`) | Resolving bugs caught in Dev / QA / Staging | `develop` / `feature/` | `develop` / `feature/` |
| **`hotfix/`** | Emergency repair for critical live production outages | `main` | `main` & `develop` |
| **`release/`** | Hardening code, version bumping, and final QA prior to rollout | `develop` | `main` & `develop` |
| **`refactor/`** | Code cleanup, modularization, or optimization without behavior change | `main` / `develop` | `main` / `develop` |
| **`chore/`** | Maintenance, updating `.gitignore`, bumping dependencies, docs | `main` / `develop` | `main` / `develop` |
| **`ci/`** (or `infra/`) | Pipeline configurations, GitHub Actions YAML, Dockerfiles | `main` / `develop` | `main` / `develop` |
| **`spike/`** (or `poc/`) | Time-boxed exploration or research; usually thrown away | `main` / `develop` | None (or rewritten to `feature/`) |

## 🔍 Deep Dive: Real-World Use Cases

### 1. `hotfix/` vs `bugfix/`
* **`hotfix/` (Urgent / Production):**
  * Spun up directly from `main` to address an active incident affecting end users (e.g., API returning 500, database deadlock).
  * Fast-tracked and merged into both `main` and `develop` to maintain synchronization.
* **`bugfix/` (Routine / Pre-Release):**
  * Cut from `develop` or your current feature branch to address issues found during testing before the code hits production.

### 2. `chore/` vs `refactor/`
* **`chore/`:** Non-business logic tasks (e.g., updating dependencies, modifying `.gitignore`, tuning linting configurations).
* **`refactor/`:** Restructuring code to improve readability, modularity, or execution efficiency without adding new features or breaking contracts.

### 3. `spike/` (Prototypes & PoCs)
* Used to evaluate new libraries, algorithms, or infrastructure options before writing production code.
* Keeps experimental or messy trial code completely isolated from clean branch histories.

## 🔄 Popular Branching Strategies

1. **GitHub Flow:**
   * Simple and lightweight.
   * Everything branches off `main` via short-lived feature branches, undergoes PR review, and merges directly back into `main`.
   * Ideal for web applications and continuous deployment environments.

2. **Git Flow:**
   * Structured model with dedicated `develop`, `feature/`, `release/`, and `hotfix/` branches.
   * Common in enterprise environments with scheduled release cycles and rigid QA sign-offs.

3. **Trunk-Based Development:**
   * Developers merge small, frequent commits into a single shared branch (`main`) multiple times a day.
   * Relies heavily on automated CI/CD suites and feature flags to manage release risk.
